## Schleifen und bedingte Anweisungen

Ein weiteres zentrales Werkzeug in Programmiersprachen sind die **Schleifen**. Eine Schleife ermöglicht es uns, bestimmte Operationen beliebig oft auszuführen. Es gibt verschiedene Arten von Schleifen, wir beschränken uns hier auf die **for-Schleifen**. Diese sind dafür gemacht, durch eine Liste durchzugehen - Element für Element. Am besten verdeutlichen wir uns das mit unserer Noten-Liste:

````python
noten = [2, 1, 3, 2, 1, 1, 6, 4, 2, 2, 5, 1]
for note in noten:
    print(note)

#Summe berechnen
summe = 0
for note in noten:
    summe = summe + note
print("Summe mit Schleife: ", summe)
print("Summe mit sum-Funktion: ", sum(noten))
````

Es gibt noch eine zweite Möglichkeit mit einer for-Schleife eine Liste durchzugehen, indem die Schleife nicht direkt über die Listenelemente läuft, sondern über die Positionen:

````python
testliste = ["element1", "element2", "element3"]

for position in [0, 1, 2]:
    print(testliste[position])

for position in range(3):
    print(testliste[position])

for position in range(len(testliste)):
    print(testliste[position])
````

Oft hat man die Situation, dass man etwas nur machen will, wenn eine bestimmte Bedingung eingetreten ist. Das kann z.B. passieren, wenn uns interessiert, wie viele Klassenarbeiten es mit der Note 2 gibt. Für so eine Aufgabe brauchen wir **bedingte Anweisungen**, die in Python mit *if* beginnen (zu deutsch "wenn"). Wenn also eine bestimmte Bedingung erfüllt ist, wird nur in diesem Fall Code ausgeführt.
Beachte: Wir vergleichen nicht mit "=", sondern mit "==" und "!=" (ungleich).

````python
eingabe = ""
if eingabe == "":
    print("Variable ist leer")
````

## Das Tic-Tac-Toe Spielfeld

Das 3x3 Spielfeld des TicTacToe kann in Tkinter am besten mit Buttons realisiert werden. Zu Beginn sind alle neun Buttons leer und zeigen "X" oder "O" an, wenn man sie einmal angeklickt hat. Erzeugen wir uns also erstmal 3 mal 3 quadratische Buttons, was wir zum Beispiel machen könnten, indem wir 9 Mal nacheinander einen Button anlegen und ihn mit `grid` positionieren. Das wäre allerdings einiges an Schreibaufwand, oder? Wir können uns die Arbeit erleichtern und for-Schleifen nutzen. Eine erste Schleife geht durch die Zeilen, eine zweite geschachtelte Schleife geht über die Spalten. Setzen wir das zunächst einmal für ein 2x2 Spielfeld um:

````python
app = Tk()
app.title("Tic Tac Toe")

for zeile in [0, 1]:
    for spalte in [0, 1]:
        btn = Button(app, text="", width=8, height=4)
        btn.grid(row=zeile, column=spalte)

app.mainloop()
````

Nanu, sollte nicht etwas passieren, wenn man auf einen Button klickt? Das stimmt, aber wir haben auch nichts gesagt, was gemacht werden sollte. Dafür müssen wir erst einmal verstehen, wie *Tkinter* ein Mausklick behandelt. Bestimmte Bedienelemente wie Buttons lösen beim Mausklick ein **Ereignis** aus. Ganz praktisch bedeudet das, dass *Tkinter* zum Zeitpunkt des Ereignisses (also z.B. im Moment des Mausklicks) eine vorgegebene Funktion ausführt. Diese Funktion beinhaltet dann alle Anweisungen, die ab dem Klick gemacht werden sollen. Den Namen der Ereignis-Funktion geben wir über den Parameter *command* an, schauen wir es uns an:

````python
def geklickt():
    print("Was passiert!")

app = Tk()
app.title("Tic Tac Toe")

button = Button(app, text="Klick mich", command=geklickt)
button.grid(row=0, column=0)

app.mainloop()
````

## Spiel gewonnen?

Machen wir uns an das wichtigste: Hat eine Person den entscheidenden Zug geamcht und das Spiel gewonnen? Zur Erinnerung: Ein TicTacToe ist gewonnen, wenn drei gleiche Symbole entweder auf einer Zeile stehen, auf einer Spalte oder in einer der beiden Diagonalen. Beginnen wir mit dem Überprüfen der ersten Zeile im folgenden Beispiel-Spielfeld:

````python
quiz = [
    ["", "", ""],
    ["X", "X", "X"],
    ["O", "", "O"]
]

#Check erste Zeile
if quiz[0][0] == quiz[0][1] == quiz[0][2]:
    print("Gewonnen!")
````

Stimmt das Ergebnis? Das Problem liegt in unserer Bedingung
```
quiz[0][0] == quiz[0][1] == quiz[0][2]
```
Natürlich sind auch drei offene Felder gleich, d.h. wir müssen zusätzlich auch noch ausschließen, dass beispielsweise das erste Feld leer ist. Die Bedingung lautet dann
```
quiz[0][0] != "" and (quiz[0][0] == quiz[0][1] == quiz[0][2])
```


#### Übung
Stelle den Code oben richtig und korrigiere die Bedingung. Ergänze den Check um die zwei weiteren Zeilen und überprüfe das Ergebnis. Baue dazu zusätzlich in der Textausgabe ein, ob "X" oder "O" gewonnen hat.

Da wir das gleiche natürlich auch noch mit den Spalten (und den Diagonalen) machen müssen, wollen wir uns den Schreibaufwand wieder etwas kleiner halten. Mit der Hilfe von *for*-Schleifen können wir den Code effizienter schreiben und die Schleife über die drei zu prüfenden Zeilen laufen lassen:

````python
for zeile in range(3):
    if (quiz[zeile][0] != "") and (quiz[zeile][0] == quiz[zeile][1] == quiz[zeile][2]):
        print("Spieler " + quiz[zeile][0] + " hat in Zeile " + str(zeile) + " gewonnen")
````

#### Übung

Nach den Zeilen und Spalten checken wir noch die beiden Diagonalen:

````python
if quiz[1][1] != "":
    if quiz[0][0] == quiz[1][1] == quiz[2][2]:
        print("Spieler " + quiz[1][1] + " hat gewonnen in der Diagonalen von links oben nach rechts unten")

    if quiz[0][2] == quiz[1][1] == quiz[2][0]:
        print("Spieler " + quiz[1][1] + " hat gewonnen in der Diagonalen von links unten nach rechts oben")
````

Für später fassen wir all das in einer **Funktion** zusammen. Eine Funktion ist eine neu definierte Anweisung mit einem eigenen Namen, die überall im Code wieder aufgerufen werden kann. Außerdem kann man der Funktion noch Parameter mitgeben. Unsere Funktion soll `ist_gewonnen` heißen, der man das auszuwertende Spielfeld übergibt:

````python
def ist_gewonnen(quiz):
    for zeile in range(3):
        if (quiz[zeile][0] != "") and (quiz[zeile][0] == quiz[zeile][1] == quiz[zeile][2]):
            return True
    for spalte in range(3):
        if (quiz[0][spalte] != "") and (quiz[0][spalte] == quiz[1][spalte] == quiz[2][spalte]):
            return True
    if quiz[1][1] != "":
        if quiz[0][0] == quiz[1][1] == quiz[2][2]:
            return True
        if quiz[0][2] == quiz[1][1] == quiz[2][0]:
            return True
    return False

print("Gewonnen? ", ist_gewonnen(quiz))
````

## Spiel unentschieden?

Mit der Funktion `ist_gewonnen()` aus dem letzten Abschnitt haben wir einen zentralen Teil unseres TocTacToe-Spiels programmiert und in "Computer-Logik" übersetzt. Jetzt widmen wir uns noch der Frage, wie der Code für den Fall "Unentschieden" aussieht, d.h. wenn alle 9 Züge gemacht worden sind und kein weiterer Zug möglich ist.

Ein möglicher Ansatz hierfür wäre, einfach alle leeren Spielfeld-Buttons zu zählen (mit zwei ineinander geschachtelten for-Schleifen). Ist die Anzahl gleich 0, ist das Spiel unentschieden. Hier ein Beispiel-Code:

````python
quiz = [
    ["X", "O", "X"],
    ["X", "O", "O"],
    ["O", "X", "X"]
]

def ist_unentschieden(quiz):
    felder_leer = 0
    for zeile in range(3):
        for spalte in range(3):
            if quiz[zeile][spalte] == "":
                felder_leer = felder_leer + 1
    if felder_leer == 0:
        return True
    else:
        return False

print("Gewonnen? ", ist_gewonnen(quiz))
print("Unentschieden? ", not(ist_gewonnen(quiz)) and ist_unentschieden(quiz))
````

## Grafische Oberfläche kommt dazu: Verbindung mit der tkinter App

Nachdem wir uns grafische Oberflächen und die Programmlogik hinter TicTacToe genauer angeschaut haben, können wir uns nun auf die Zielgerade begeben und alles zu einer TicTacToe App zusammenbauen. Allerdings, eine Sache fehlt noch: Die Ereignis-Funktion, die ausgeführt wird, wenn ein leerer Spielfeld-Button angeklickt wird.

Klickt eine Person z.B. auf den Spielfeld-Button in der n-ten Zeile und m-ten Spalte, so muss das Spielfeld aktualisiert werden und entweder ein "X" oder ein "O" an diese Position eingefügt werden:
```
quiz[n][m] = aktueller_spieler
```
Danach muss in der Ereignis-Funktion mit `ist_gewonnen(quiz)` und `ist_unentschieden(quiz)` nur noch entschieden werden, ob die aktuell spielende Person mit ihrem Zug gewonnen hat oder das Spielfeld nun unentschieden ist.

Da wir für das Aktualisieren des Spielfeldes wissen müssen, wer als nächstes dran ist ("X" oder "O"), müssen wir uns noch um die Variable `aktueller_spieler` kümmern. Den nächsten Spieler können wir einfach bestimmen, indem wir im Spielfeld `quiz` alle gemachten Züge zählen. Ist die Anzahl der Züge gerade, spielt "X" als nächstes; ist die Anzahl ungerade, spielt "O". Der nächste Code zeigt eine mögliche Implementierung dessen:

````python
quiz = [
    ["", "", ""],
    ["X", "X", "X"],
    ["O", "", "O"]
]

anzahl_zuege = 9 - quiz[0].count("") - quiz[1].count("") - quiz[2].count("")
aktueller_spieler = ["X", "O"][anzahl_zuege % 2]  #anzahl_zuege % 2 ist der Rest, den die Division durch 2 lässt
print("Als nächstes ist dran: ", aktueller_spieler)
````

Unsere Überlegungen zur Ereignis-Funktion können wir nun praktisch umsetzen und eine erste vorläufige Version der TicTacToe-App programmieren. Im nächsten Code-Block findest du das Grundgerüst der App - die fertige Tkinter-Oberfläche mit den beiden Ereignissen `clicked_spielfeld()` und `clicked_neu()`, die ausgelöst werden, wenn ein Spielfeld-Button bzw. der "Neues Spiel"-Button angeklickt wird:

````python
from tkinter import *  


def clicked_spielfeld(zeile, spalte, buttons, spielfeld, spieler):
    anzahl_zuege = 9 - spielfeld[0].count("") - spielfeld[1].count("") - spielfeld[2].count("")
    aktueller_spieler = spieler[anzahl_zuege % 2]

    buttons[zeile][spalte].configure(text = aktueller_spieler)
    spielfeld[zeile][spalte] = aktueller_spieler

def clicked_neu(buttons, spielfeld):
    for zeile in range(3):
        for spalte in range(3):
            buttons[zeile][spalte].configure(text = "")
            spielfeld[zeile][spalte] = ""


# grafische Benutzeroberfläche erstellen...
app = Tk()
app.title("Tic Tac Toe")
app.resizable(0,0)

# initialisiere alle Listen
buttons = [[None] * 3, [None] * 3, [None] * 3]
spielfeld = [[""] * 3, [""] * 3, [""] * 3]
spieler = ["X", "O"]

# erzeuge Buttons
for i in range(3):
    for j in range(3):
        buttons[i][j] = Button(app,
            height = 4,
            width = 8,
            font = ("Helvetica", "20"),
            command = lambda z=i, s=j, b=buttons, f=spielfeld, p=spieler: clicked_spielfeld(z, s, b, f, p)
        )
        buttons[i][j].grid(row = i, column = j)
# erstelle "Neues Spiel"-Button
btn_neu = Button(app, text = "Neues Spiel", command = lambda b=buttons, f=spielfeld: clicked_neu(b, f))
btn_neu.grid(row = 1, column = 3)

app.mainloop()
````

Zur Fertigstellung unserer App müssen wir `clicked_spielfeld()` noch so ergänzen, dass eine Meldung auf dem Bildschirm erscheint, ob die letzte Person am Zug das Spiel gewonnen hat oder das Spielfeld nun unentschieden ist. Die Vorarbeit dazu haben wir schon geleistet, denn wir können einfach unsere beiden Funktionen `ist_gewonnen()` und `ist_unentschieden` von weiter oben wiederverwenden. Die Ausgabe, ob das Spiel gewonnen oder unentschieden ist, packen wir in ein Dialog-Fenster. Das geht am einfachsten mit einer tkinter-**Messagebox**:
```
from tkinter import messagebox
```

Fügen wir nun alles zusammen, haben wir eine fertige und funktionierende TicTacToe-App. Schau dir dazu den nächsten Code-Block an, der die gesamte App umfasst. Bevor du das erste Spiel spielst, verschaffe dir zunächst nochmal einen Überblick über den finalen Code der App.

````python
from tkinter import *  
from tkinter import messagebox


def ist_gewonnen(quiz):
    for zeile in range(3):
        if (quiz[zeile][0] != "") and (quiz[zeile][0] == quiz[zeile][1] == quiz[zeile][2]):
            return True
    for spalte in range(3):
        if (quiz[0][spalte] != "") and (quiz[0][spalte] == quiz[1][spalte] == quiz[2][spalte]):
            return True
    if quiz[1][1] != "":
        if quiz[0][0] == quiz[1][1] == quiz[2][2]:
            return True
        if quiz[0][2] == quiz[1][1] == quiz[2][0]:
            return True
    return False

def ist_unentschieden(quiz):
    felder_leer = quiz[0].count("") + quiz[1].count("") + quiz[2].count("")
    if felder_leer == 0:
        return True
    else:
        return False

def clicked_spielfeld(zeile, spalte, buttons, spielfeld, spieler):
    # Button bereits geklickt?
    if buttons[zeile][spalte].cget("text") != "":
        return

    anzahl_zuege = 9 - spielfeld[0].count("") - spielfeld[1].count("") - spielfeld[2].count("")
    aktueller_spieler = spieler[anzahl_zuege % 2]

    buttons[zeile][spalte].configure(text = aktueller_spieler)
    spielfeld[zeile][spalte] = aktueller_spieler

    if ist_gewonnen(spielfeld):
        messagebox.showinfo("Winner", aktueller_spieler + " hat gewonnen!")
    elif ist_unentschieden(spielfeld):
        messagebox.showinfo("Unentschieden", "Unentschieden!")
    else:
        return
    for zeile in range(3):
        for spalte in range(3):
            buttons[zeile][spalte].configure(state = DISABLED)

def clicked_neu(buttons, spielfeld):
    for zeile in range(3):
        for spalte in range(3):
            buttons[zeile][spalte].configure(state = NORMAL)
            buttons[zeile][spalte].configure(text = "")
            spielfeld[zeile][spalte] = ""


# grafische Benutzeroberfläche erstellen...
app = Tk()
app.title("Tic Tac Toe")
app.resizable(0,0)

# initialisiere alle Listen
buttons = [[None] * 3, [None] * 3, [None] * 3]
spielfeld = [[""] * 3, [""] * 3, [""] * 3]
spieler = ["X", "O"]

# erzeuge Buttons
for i in range(3):
    for j in range(3):
        buttons[i][j] = Button(app,
            height = 4,
            width = 8,
            font = ("Helvetica", "20"),
            command = lambda z=i, s=j, b=buttons, f=spielfeld, p=spieler: clicked_spielfeld(z, s, b, f, p)
        )
        buttons[i][j].grid(row = i, column = j)
# erstelle "Neues Spiel"-Button
btn_neu = Button(app, text = "Neues Spiel", command = lambda b=buttons, f=spielfeld: clicked_neu(b, f))
btn_neu.grid(row = 1, column = 3)

app.mainloop()
````
